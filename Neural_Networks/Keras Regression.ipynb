## Scaling and Train Test Split

```python
X = df.drop('price', axis=1).values  # Features
y = df['price'].values  # Target variable
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=101)  # Split the dataset into training and testing sets
```

## Scaling

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()  # Initialize the MinMaxScaler
```

```python
X_train = scaler.fit_transform(X_train)  # Fit the scaler to the training data and transform it
```

```python
X_test = scaler.transform(X_test)  # Transform the test data using the same scaler
```

```python
X_train.shape
```

```python
X_test.shape
```

## Creating Model

```python
import tensorflow as tf
```

```python
from keras.models import Sequential
from keras.layers import Dense,Activation
from keras.optimizers import Adam
```

```python
model = Sequential()  # Initialize the neural network model
model.add(Dense(19,activation='relu'))
model.add(Dense(19,activation='relu'))
model.add(Dense(19,activation='relu'))
model.add(Dense(19,activation='relu'))
model.add(Dense(1))

model.compile(optimizer='adam', loss='mse')  # Compile the model with Adam optimizer and mean squared error loss
```

## Training the  model

```python
model.fit(x=X_train, y=y_train,
           validation_data=(X_test, y_test),
             batch_size=128, epochs=400)  # Train the model on the training data
```

```python
losses = pd.DataFrame(model.history.history)  # Convert the training history to a DataFrame
```

```python
losses.plot()  # Plot the training and validation loss over epochs
```

## Evaluation on Test Data

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, explained_variance_score
```
